```python
import os
import pandas as pd

# Thư mục chứa các file .csv
csv_folder = "/home/datnvt/project/Medical_CLARA/data_detection"

# Lấy danh sách tất cả các file .csv trong thư mục
csv_files = [f for f in os.listdir(csv_folder) if f.endswith(".csv")]

# Gộp tất cả lại
dfs = [pd.read_csv(os.path.join(csv_folder, f)) for f in csv_files]
combined_df = pd.concat(dfs, ignore_index=True)

# Lưu lại file đã gộp
combined_df.to_csv("/home/datnvt/project/Medical_CLARA/data_detection/xray_detection_combined.csv", index=False)

print(f"Đã gộp {len(csv_files)} file -> 'xray_detection_combined.csv'")
```
